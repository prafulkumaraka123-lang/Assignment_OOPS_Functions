## Question 1
**What is a function in Python? Explain how functions are created and called. Also explain parameters, arguments, default parameters, `*args`, and `**kwargs` with suitable examples.**

### Answer
A **function** is a reusable block of code designed to perform a specific task. Functions are created using the `def` keyword and are called using the function name followed by parentheses.

- **Parameter:** A variable listed in the function definition.
- **Argument:** The actual value passed to a function when it is called.
- **Default parameter:** A parameter that has a predefined value.
- **`*args`:** Allows a function to accept any number of positional arguments.
- **`**kwargs`:** Allows a function to accept any number of keyword arguments.

Example:
```python
def add(a, b):
    return a + b

print(add(10, 20))

def greet(name="Student"):
    print(f"Hello {name}")

def total(*args):
    return sum(args)

def details(**kwargs):
    for key, value in kwargs.items():
        print(key, value)
```

````python
# Question 1 - Example
def add(a, b):
    return a + b

print("Addition:", add(10, 20))

def greet(name="Student"):
    print(f"Hello {name}")

greet()
greet("Praful")

def total(*args):
    return sum(args)

print("Total using *args:", total(10, 20, 30))

def details(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")

details(name="Praful", course="B.Tech")
````

## Question 2
**Create a Python program with a Person class containing the person's name and a method to display it. Then create a Student class that inherits from Person and adds the student's course. Create an object of Student and display both details.**

### Answer
This program demonstrates **class creation, objects, inheritance, attributes, and methods**. `Student` inherits the `name` attribute and `display_name()` method from `Person`, and adds its own `course` attribute.

````python
# Question 2
class Person:
    def __init__(self, name):
        self.name = name

    def display_name(self):
        print("Name:", self.name)

class Student(Person):
    def __init__(self, name, course):
        super().__init__(name)
        self.course = course

    def display_details(self):
        self.display_name()
        print("Course:", self.course)

student = Student("Praful Kumar", "B.Tech")
student.display_details()
````

## Question 3
**What is variable scope in Python? Explain local and global variables. Why should excessive use of the global keyword be avoided? Also explain nested functions and closures.**

### Answer
**Variable scope** refers to the region of a program where a variable can be accessed.

A **local variable** is created inside a function and is normally accessible only inside that function.

A **global variable** is created outside a function and can be accessed throughout the program/module.

The `global` keyword allows a function to modify a global variable. Excessive use of `global` should be avoided because it can make programs harder to understand, test, maintain, and debug.

A **nested function** is a function defined inside another function.

A **closure** is a nested function that remembers and can access a variable from its enclosing function even after the enclosing function has finished.

````python
# Question 3 - Examples
x = 100  # global variable

def outer():
    y = 20  # local variable
    def inner():
        return y  # closure accesses y from outer()
    return inner

show_value = outer()
print("Closure result:", show_value())
print("Global variable:", x)
````

## Question 4
**What is a lambda function? Explain how `map()`, `filter()`, and `reduce()` work in Python with suitable examples.**

### Answer
A **lambda function** is a small anonymous function written using the `lambda` keyword. It can take arguments and contains a single expression.

- **`map()`** applies a function to every item in an iterable.
- **`filter()`** selects items for which a condition is true.
- **`reduce()`** repeatedly combines elements to produce one result. It is available from the `functools` module.

````python
# Question 4
from functools import reduce
numbers = [1, 2, 3, 4]
square = lambda x: x * x
squares = list(map(square, numbers))
even_numbers = list(filter(lambda x: x % 2 == 0, numbers))
total = reduce(lambda a, b: a + b, numbers)
print("Squares using map():", squares)
print("Even numbers using filter():", even_numbers)
print("Sum using reduce():", total)
````

## Question 5
**Look at the code below and answer the questions: What will be printed? Why is the Parent version of `show()` not called?**

### Answer
The program prints:

```text
Child
```

The `Child` class **overrides** the `show()` method inherited from `Parent`. When `obj.show()` is called, Python uses the method defined in the object's class (`Child`). Therefore, the `Parent` version is not called automatically. The `Parent` version could be called explicitly using `super().show()` inside the child method.

````python
# Question 5 - Code from the assignment
class Parent:
    def show(self):
        print("Parent")

class Child(Parent):
    def show(self):
        print("Child")

obj = Child()
obj.show()
````

## Question 6
**Create a custom iterator class called `EvenNumbers` that generates even numbers from 2 to 10. Use `__iter__()`, `__next__()`, and `StopIteration`.**

````python
# Question 6
class EvenNumbers:
    def __init__(self):
        self.current = 2

    def __iter__(self):
        return self

    def __next__(self):
        if self.current <= 10:
            value = self.current
            self.current += 2
            return value
        raise StopIteration

even_numbers = EvenNumbers()
print("Even numbers from 2 to 10:")
for number in even_numbers:
    print(number)
````

## Question 7
**Write a Python program using multiple inheritance. Create two classes, `Teacher` and `Coder`, with one method in each. Create a `Trainer` class that inherits from both and use an object of `Trainer` to call both methods.**

### Answer
**Multiple inheritance** means that one class inherits from more than one parent class. Here, `Trainer` inherits from both `Teacher` and `Coder`, so a `Trainer` object can access methods from both classes.

````python
# Question 7
class Teacher:
    def teach(self):
        print("Teacher is teaching.")

class Coder:
    def code(self):
        print("Coder is writing code.")

class Trainer(Teacher, Coder):
    pass

trainer = Trainer()
trainer.teach()
trainer.code()
````

## Question 8
**What is polymorphism? Explain it in your own words with a Python example where two different classes have the same method name but perform different tasks.**

### Answer
**Polymorphism** means "many forms." In Python, different classes can have a method with the same name, but each class can implement that method differently. The same method call can therefore produce different behavior depending on the object.

````python
# Question 8
class Dog:
    def sound(self):
        print("Dog says: Woof")

class Cat:
    def sound(self):
        print("Cat says: Meow")

dog = Dog()
cat = Cat()
dog.sound()
cat.sound()
````

**Consider the list `numbers = [1, 2, 3, 4, 5, 6]`. Use `map()` to calculate the square of every number, `filter()` to select only even numbers, and `reduce()` to calculate the sum of all numbers. Display all three results.**

## Question 10
**You are developing a payment system for an online shopping application. The system should support UPI, credit card, and wallet payments. Each payment type should have its own way of processing the payment, but the main program should be able to call the same `pay()` method for all of them. Design a simple Python solution using inheritance and polymorphism, and explain how it works.**

### Answer
The solution uses a common parent class called `Payment` with a `pay()` method. The classes `UPIPayment`, `CreditCardPayment`, and `WalletPayment` inherit from it and provide their own implementation of `pay()`. The main program can store different payment objects and call the same `pay()` method on each object. Python then executes the appropriate method based on the object's actual class. This demonstrates **inheritance and polymorphism**.

````python
# Question 10
class Payment:
    def pay(self, amount):
        print(f"Processing payment of ₹{amount}")

class UPIPayment(Payment):
    def pay(self, amount):
        print(f"Paid ₹{amount} using UPI")

class CreditCardPayment(Payment):
    def pay(self, amount):
        print(f"Paid ₹{amount} using Credit Card")

class WalletPayment(Payment):
    def pay(self, amount):
        print(f"Paid ₹{amount} using Wallet")

payments = [UPIPayment(), CreditCardPayment(), WalletPayment()]
for payment in payments:
    payment.pay(1000)
````
